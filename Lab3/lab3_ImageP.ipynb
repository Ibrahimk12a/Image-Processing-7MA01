# Lab 3:   Image Manipulations using OpenCV Library

### Task 1: Geometric Transformations
In this task, I apply geometric transformations to the image: increasing its size, rotating it by 120 degrees, and performing a sheer operations.


```python
import cv2
import numpy as np

img = cv2.imread('images/cameraman.tif')
rows, cols = img.shape[:2]

resized_img = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)

M_rot = cv2.getRotationMatrix2D((cols/2, rows/2), 120, 1)
rotated_img = cv2.warpAffine(img, M_rot, (cols, rows))

M_shear = np.float32([[1, 0.5, 0], [0, 1, 0]])
sheared_img = cv2.warpAffine(img, M_shear, (int(cols*1.5), rows))

cv2.imshow('Original', img)
cv2.imshow('Resized', resized_img)
cv2.imshow('Rotated', rotated_img)
cv2.imshow('Sheared', sheared_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Task 2: Intensity Transformations
In this task, I apply intensity transformations to create a negative image, apply a log transformation, and use a power law transformation to adjust contrast.

```python
negative_img = 255 - img

max_val = float(np.max(img))
c = 255 / (np.log(1 + max_val))
log_img = np.array(c * (np.log(1 + np.float32(img))), dtype=np.uint8)

# Gamma > 1 increases contrast
gamma = 2.0 
power_law_img = np.array(255 * (img / 255.0) ** gamma, dtype='uint8')

cv2.imshow('Negative Image', negative_img)
cv2.imshow('Log Transform', log_img)
cv2.imshow('Power Law', power_law_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
