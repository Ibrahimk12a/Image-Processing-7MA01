# Lab 4:   Intensity Transformations and Filtering-Spatial Domain

### Task 1: Rescale Intensity Values (3rd and 80th percentiles)

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure

img_moon = data.moon()

p3, p80 = np.percentile(img_moon, (3, 80))
img_rescale = exposure.rescale_intensity(img_moon, in_range=(p3, p80))

fig1 = plt.figure(figsize=(10, 7))
fig1.add_subplot(2, 1, 1)
plt.imshow(img_rescale, cmap='gray')
plt.axis('off')
plt.title('Contrast stretch')

fig1.add_subplot(2, 1, 2)
plt.hist(img_rescale.flat, bins=256, range=(0, 255))
plt.xlabel('gray level values')
plt.ylabel('Number of pixels')
plt.show()
```

### Task 2: Histogram Equalization


```python
img_eq = exposure.equalize_hist(img_moon)

fig2 = plt.figure(figsize=(10, 7))
fig2.add_subplot(2, 1, 1)
plt.imshow(img_eq, cmap='gray')
plt.axis('off')
plt.title('Equalized Image')

fig2.add_subplot(2, 1, 2)
plt.hist(img_eq.flat, bins=256, range=(0, 1))
plt.xlabel('gray level values')
plt.ylabel('Number of pixels')
plt.show()
```

### Task 3: Histogram Matching

```python
from skimage.exposure import match_histograms

reference = data.rocket()
image = data.chelsea()

matched = match_histograms(image, reference, channel_axis=-1)

fig3, axes = plt.subplots(nrows=1, ncols=3, figsize=(15, 5))

axes[0].imshow(image)
axes[0].set_title('Source Image (Chelsea)')
axes[0].axis('off')

axes[1].imshow(reference)
axes[1].set_title('Reference Image (Rocket)')
axes[1].axis('off')

axes[2].imshow(matched)
axes[2].set_title('Matched Image')
axes[2].axis('off')

plt.show()
```
