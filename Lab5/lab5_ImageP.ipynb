# Lab 5:   Spatial Filtering – Smoothing and Sharpening

### Task 1: Convolve an image with a 7x7 box filter

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('images/chelsea.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

smoothed_image = cv2.blur(image, (7, 7))

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(image)
ax[0].axis('off')
ax[1].imshow(smoothed_image)
ax[1].axis('off')
plt.show()
```

### Task 2: Apply a 5x5 Gaussian filter, then a 21x21 Gaussian filter


```python
image = cv2.imread('images/chelsea.png')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gaussian_5x5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21x21 = cv2.GaussianBlur(image, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title('Original Image')
ax[0].axis('off')
ax[1].imshow(gaussian_5x5)
ax[1].set_title('Gaussian 5x5')
ax[1].axis('off')
ax[2].imshow(gaussian_21x21)
ax[2].set_title('Gaussian 21x21')
ax[2].axis('off')
plt.show()
```

### Task 3: Apply a 3x3 Laplacian filter to sharpen the image

```python
image = cv2.imread('../Lab5/images/cameraman.tif', cv2.IMREAD_GRAYSCALE)

image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

sharpened = np.clip(image_float - laplacian, 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image_float, cmap='gray')
ax[0].set_title('Original Image')
ax[0].axis('off')
ax[1].imshow(laplacian, cmap='gray')
ax[1].set_title('Laplacian Filter')
ax[1].axis('off')
ax[2].imshow(sharpened, cmap='gray')
ax[2].set_title('Sharpened Image')
ax[2].axis('off')
plt.show()
```
